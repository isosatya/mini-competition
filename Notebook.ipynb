```python
# Load the labels data
labels_df = pd.read_csv('data/raw/Training_Data_Labels.csv')

# Separate labels based on city
y_sj = labels_df[labels_df['city'] == 'sj']['total_cases']
y_iq = labels_df[labels_df['city'] == 'iq']['total_cases']
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Function to convert weekofyear to month
def week_to_month(week):
    return (week // 4) + 1  # Approximate conversion: 4 weeks ≈ 1 month

# Create subplots
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 5))

# Plot for San Juan
sns.histplot(data=df_sj, x='weekofyear', weights=labels_df[labels_df['city'] == 'sj']['total_cases'], 
             bins=12, ax=ax1)
ax1.set_title('San Juan (sj) - Total Cases by Month')
ax1.set_xlabel('Month')
ax1.set_ylabel('Total Cases')
ax1.set_xticks(range(1, 13))
ax1.set_xticklabels(['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 
                     'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec'])

# Plot for Iquitos
sns.histplot(data=df_iq, x='weekofyear', weights=labels_df[labels_df['city'] == 'iq']['total_cases'], 
             bins=12, ax=ax2)
ax2.set_title('Iquitos (iq) - Total Cases by Month')
ax2.set_xlabel('Month')
ax2.set_ylabel('Total Cases')
ax2.set_xticks(range(1, 13))
ax2.set_xticklabels(['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 
                     'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec'])

plt.tight_layout()
plt.show()
```
